    - Chosen dataset:
A non-store online retail that focuses on gift-ware and is based in the UK, and that company's transaction history between 1st of December, 2009 until 9th of December, 2011.

    - Business problem:
To which extent are the customers willing to buy another product from Online Retail II within 60 days of another purchase, hence showcasing their loyalty to the company?

    - Why is it important?:
Marketing uses a lot more resources for various store perks, such as coupon codes. These perks often get used up by customers that may either never make another purchase from this company or by customers that would come back no matter if they got that perk or not.
Due to this, it is very important to analyse how often customers buy products from this company as well as the amount of time that passes between each purchase, to determine their 'loyalty'. 

    - Benefits:
Being able to identify the 'loyalty' of a customer is very significant, as that way, marketing would know where to delegate its resources more effectively.
In addition to perks becoming more targeted, as the customers are being paid attention to more closely, this would most likely to a general increase in the customer base, as more people would end up buying from this company.

    - How will we know if its beneficial or not?
A way to do this would be to constantly update the database and comparing the results. Some examples would be, if the changes made in marketing are effective, there might be either more customers returning back to buy from this company or just an overall higher number of customers than what the online retail started with.

    - How is the relevant data being collected?
For this assignment, I am using a dataset. Although in an actual setting, the company would first start by collecting the information regarding the purchases. The details of the transaction history would focus on the same features as the columns in the dataset that I am using. 
To add on, the company could also send out surveys with additional bonuses that focuses on collecting the customer's feelings, thus testing their loyalty and likeness to come back to make another purchase soon. This type of engagement does not necessarily need to buy a survey, but it needs to be able to collect the relevant information.

    - Formulating this problem as a machine learning task:
This problem uses the idea of a binary classification task, which takes the customer purchase history before the cutoff date and assigns them the label 1 if they made another purchase in the 60 days after the cutoff. If not, they are assigned the label 0. The model learns the patterns as well as the relationship between the purchasing behaviour and the possibility of the customer coming back within that time frame. Such a model hence is able to turn 'loyalty' into a metric that is not only predictable, but also measurable by specific features, allowing marketing to have a clearer understanding of their customer base and to minimize misallocating the costs.
